**SOLUCIÓN.** El espacio muestral tanto para $S_n$ como para $D_n$ es lo mismo:

 $\Omega_1=\{1,2,3,4,5,6\}\times\{1,2,3,4,5,6\}$

Ahora, vemos el rango de cada evento:
* $R_{S,n}$ ={1,2,3,4,5,6} X {1,2,3,4,5,6}
* $R_{D,n}$ ={0,1,2,3,4,5}








```python
# Espacio muestral
from itertools import product
Omega1 = set(product({1,2,3,4,5,6}, repeat =2))
print(Omega1)
print("La cardinalidad de Omega1 es:", len(Omega1))
```

La descripción conjunta de los eventos en cuestión son:

$S_n=\{(i,j)\in\Omega_1:\ i+j=n\}$

y

$D_m=\{(i,j)\in\Omega_1:\ |i-j|\geq m\}$

```python
 # La suma de los dos dados en n
def S(n):
    S = {o for o in Omega1 if o[0]+o[1] == n}
    return S
```

Veamos los eventos para cada 𝑛:

```python
Rango_Sn = range(2,13)
Eventos_Sn = {n : S(n) for n in Rango_Sn }
Eventos_Sn
```

Similarmente,

```python
# La suma de los dos dados en n
def D(m):
    D = {o for o in Omega1 if abs(o[0]-o[1]) >= m}
    return D
```

¿Los eventos son independientes?

```python
def Indep(A,B):
    indep = P(A.intersection(B), Omega1)==P(A, Omega1)*P(B,Omega1)
    return indep
```

```python
Independencia = [(n,m) for (n,m) in Rango if Indep(D(m),S(n))==True]
Independencia
```

Por lo tanto, habrá independencia cuando $n\in R_{S_n}$ y $m=0$.

**Ejercicio 3.**


Una urna contiene 10 bolas numeradas del 1 al 10. Un experimento consiste en elegir, al azar y sin reemplazo, dos bolas de la urna, consecutivamente.

Calcule:

**a)** La probabilidad de que la suma de los dos números elegidos sea par.

**b)** La probabilidad de que en la primera elección resulte par dado que la suma fue par.

**Solución.**

Sean:

- $B$ = la suma de los números elegidos es par.
- $A_1$ = el número de la primera elección es par.
- $A_2$ = el número de la segunda elección es impar.

```python
Omega2 = {i for i in range(1,11)}
B = {(i,j) for i in Omega2 for j in Omega2 if (i+j)%2==0}
A1 = {i for i in Omega2 if i%2==0}
A2 = {i for i in Omega2 if i%2==1}
```

Calculemos $\mathbb{P}(B\mid A_1)$ y $\mathbb{P}(B\mid A_2)$

```python
BA1 = {(i, j) for i in A1 for j in Omega2 if (i+j)%2==0 and i!=j}
BA1
```

Así, $\mathbb{P}(B\mid A_1)=\frac{4}{9}=\mathbb{P}(B\mid A_2)$.

Por lo tanto, $\mathbb{P}(B)=\frac{4}{9}$.
